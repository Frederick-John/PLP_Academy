## Import Libraries

```python
# Data analysis and visualization libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure plots display nicely in notebooks
%matplotlib inline
```

# Task 1: Load and Explore Dataset

For this assignment, let’s use the Iris dataset (a classic dataset in machine learning).

```python
from sklearn.datasets import load_iris

# Load dataset from sklearn
iris_data = load_iris(as_frame=True)
df = iris_data.frame  # Convert to DataFrame

# Display first rows
print("First 5 rows of the dataset:")
display(df.head())

# Inspect structure
print("\nDataset Info:")
print(df.info())

# Check for missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

✅ Explanation:

- .head() → first rows to inspect.

- .info() → data types and structure.

- .isnull() → ensures dataset has no missing values.

Since Iris is clean, no cleaning is required. But let’s add an error handling example if a CSV is loaded:

```python
try:
    df = pd.read_csv("data.csv")
except FileNotFoundError:
    print("Error: The file was not found. Please check the filename or path.")
```

## Task 2: Basic Data Analysis

```python
# Basic statistics
print("\nDescriptive Statistics:")
display(df.describe())

# Grouping example: mean values per species
grouped = df.groupby("target").mean()
print("\nMean of numerical features by species:")
display(grouped)
```

✅ Explanation:

- .describe() → mean, std, min, max, quartiles.

- .groupby() → compare features across species (setosa, versicolor, virginica).

## Task 3: Data Visualization

### a) Line Chart (Trend Example)

Let’s create a fake time series by assuming sample IDs represent days.

```python
plt.figure(figsize=(10, 6))
plt.plot(df.index, df["sepal length (cm)"], label="Sepal Length")
plt.title("Line Chart: Sepal Length Over Samples")
plt.xlabel("Sample Index")
plt.ylabel("Sepal Length (cm)")
plt.legend()
plt.show()
```

### b) Bar Chart (Comparison Across Species)

```python
plt.figure(figsize=(8, 6))
sns.barplot(x="target", y="sepal length (cm)", data=df, estimator="mean", errorbar=None)
plt.title("Average Sepal Length per Species")
plt.xlabel("Species")
plt.ylabel("Average Sepal Length (cm)")
plt.show()
```

### c) Histogram (Distribution of a Numerical Column)

```python
plt.figure(figsize=(8, 6))
plt.hist(df["petal length (cm)"], bins=20, color="skyblue", edgecolor="black")
plt.title("Histogram of Petal Length")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Frequency")
plt.show()
```

### d) Scatter Plot (Relationship Between Two Variables)

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x="sepal length (cm)", y="petal length (cm)", hue="target", data=df, palette="Set1")
plt.title("Scatter Plot: Sepal Length vs Petal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")
plt.legend(title="Species")
plt.show()
```

## Findings & Observations

1. Descriptive Statistics show that petal measurements vary more than sepal measurements.

2. Grouped Means confirm clear differences between species: setosa has significantly smaller petal lengths.

3. Line Chart shows general stability in sepal length across samples.

4. Bar Chart highlights that virginica species generally have the longest sepals.

5. Histogram reveals a bimodal distribution for petal length (clusters for setosa vs others).

6. Scatter Plot demonstrates strong correlation between sepal length and petal length, with clusters clearly separating species.
